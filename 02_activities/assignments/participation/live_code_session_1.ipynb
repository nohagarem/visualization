## Making a basic figure with matplotlib

```python
# load libraries

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)
```

```python
fig, ax = plt.subplots(figsize=(5,3)) 
# width of the figure is 5 inches and height is 3 inches
ax.scatter(x,y)
```

```python
# change to a bar graph
fig, ax = plt.subplots(figsize=(5,3)) 
ax.bar(x,y)
```

```python
# change to a horizontal bar chart
fig, ax = plt.subplots(figsize=(5,3)) 
ax.barh(x,y)
```

```python
# change to a line plot
fig, ax = plt.subplots(figsize=(5,3)) 
ax.plot(x,y)
```

```python
# change to a histogram
fig, ax = plt.subplots(figsize=(5,3))
ax.hist(y) # we don't need the x values here, we're counting how many y values are in each bin.
```

```python
# first define the plot then add the labels and title

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
# changing the font\color\size
# we make a dictionary with the properties we want

font1 = {'family': 'sans-serif', 'color':'blue', 'size':20}
font2 = {'family': 'monospace', 'color':'green', 'size':14}
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict=font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
fig.tight_layout()
```

```python
# we can move our labels/title with the loc argument
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict=font1, loc='left')
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
fig.tight_layout()
```

```python
# we can also modify the data points in the scatter plot

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,
           y,
           marker='*',
           color='indigo')
#fig.show()
```

```python
# and we can adjust the style and width of the line connecting the points in a line plot

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,
        y,
        marker='*',
        color='indigo',
        linestyle = '--',
        linewidth = 2)
```

```python
# we can use color names or use the hex code of the color

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,
        y,
        marker='*',
        color='#7425b9',
        linestyle = '--',
        linewidth = 2)
```

```python
# we can customize the marker as well

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,
        y,
        marker='*',
        markersize = 12,
        color='#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#7425b9')
```

```python
# we can also add grid lines

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,
           y,
           marker='*',
           color='indigo')
ax.grid(axis='y')
```

```python
# we can customize the appearance of the grid lines

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,
           y,
           marker='*',
           color='indigo')
ax.grid(axis='y', color = 'blue', linewidth = 2, linestyle = '-.')
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,
           y,
           marker='*',
           color='indigo')
ax.grid(axis='y', color = 'blue', linewidth = 2, linestyle = '-.')
ax.grid(axis = 'x', color = 'red', linewidth = 1, linestyle = '--')
```

```python
# a contour plot with seaborn

import seaborn as sns
```

```python
# load the iris dataset
df = sns.load_dataset('iris')
```

```python
# create a 2D density plot
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length)
plt.show()
```

```python
# we can add colormap to the plot
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length, cmap="Blues", fill=True)
plt.show()
```

```python
# we can also adjust the smoothing parameter, bw_adjust: the higher the value, the smoother the plot
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length, cmap="Blues", fill=True, bw_adjust=0.5)
plt.show()
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length, cmap="Blues", fill=True, bw_adjust = 2)
plt.show()
```
