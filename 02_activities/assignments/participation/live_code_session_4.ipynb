```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# define our sample data

np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y2 = np.array([110, 180, 240, 99, 220])

```

```python
# define two sets of axes and their relative positions
fig, (ax1, ax2) = plt.subplots(ncols = 2, #two columns
                               nrows = 1, # 1 row
                               figsize = (7,3)) # 7inches wide x 3inches high
ax1.scatter(x1,y1)  # first column
ax2.bar(x2,y2)      # second column
```

```python
fig, (ax1, ax2) = plt.subplots(ncols = 2, #two columns
                               nrows = 1, # 1 row
                               figsize = (7,3)) # 7inches wide x 3inches high
ax1.scatter(x1,y1, marker = '*', 
        color='black')  # first column
ax1.set_title('scatter plot')
ax2.bar(x2,y2, color='gray')      # second column
ax2.set_title('Bar Chart')
```

```python
# if we don't have a grid arrangement
# we'll have labels for the axes

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4))
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
```

```python
# if we don't have a grid arrangement
# we'll have labels for the axes

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4))
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes['ax1'].set_xlabel('A Big Label', fontsize = 18)
someaxes['ax2'].set_xlabel('Another Label', fontsize = 18)
someaxes['ax3'].set_xlabel('Label2: 2 Fast 2 Furious', fontsize = 18)
```

```python
# to make sure everything we add to the graph is not cut off we need tight_layout, or contrained

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize = (7,4),
                                    layout = 'constrained')
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes['ax1'].set_xlabel('A Big Label', fontsize = 18)
someaxes['ax2'].set_xlabel('Another Label', fontsize = 18)
someaxes['ax3'].set_xlabel('Label2: 2 Fast 2 Furious', fontsize = 18)
```

```python
# we want to make 2 visuals on the same axes

x = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize=(7,3))

ax.bar(x, y1, color = 'indigo')
ax.plot(x, y2, color = 'red')
```

```python
# calculate standard deviation

y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2, color='red')
```

```python
# add error info by using errorbar()
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2, color='red')

ax.errorbar(x, y2, yerr = y2_sd, fmt='none')
```

```python
# customize errorbar appearance
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2, color='red')

ax.errorbar(x, 
            y2, 
            yerr = y2_sd, 
            fmt='none',
            ecolor = 'indigo',
            elinewidth = 4,
            capsize = 6,
            capthick = 4)
```

```python
# if we don't want to see the errors for every single point, we can do errorevery specified interval
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2, color='red')

ax.errorbar(x, 
            y2, 
            yerr = y2_sd, 
            fmt='none',
            ecolor = 'indigo',
            elinewidth = 4,
            capsize = 6,
            capthick = 4,
            errorevery = 2)
```

```python
# How to add images to a plot!!!

# load required libraries

from PIL import Image # open images
import requests # to get images from URLs
from io import BytesIO # to store images
```

```python
# get image from the internet

response = requests.get('https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png')
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2, color = 'red')

# overlay a new axis, ax_image, on our figure on top of 'ax' to act as a container for our image
ax_image = fig.add_axes( [0.1, # x coordinate (on figure, not axes)
                          0.11, # y coordinate (on figure, not axes)
                          0.15, # image width
                          0.35 # image height
                          ])
```

```python
# now add the image

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2, color = 'red')

# overlay a new axis, ax_image, on our figure on top of 'ax' to act as a container for our image
ax_image = fig.add_axes( [0.1, # x coordinate (on figure, not axes)
                          0.11, # y coordinate (on figure, not axes)
                          0.15, # image width
                          0.35 # image height
                          ])

ax_image.imshow(image)
ax_image.axis('off')
```
