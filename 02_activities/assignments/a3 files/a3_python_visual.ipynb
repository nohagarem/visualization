```python
# load libraries

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# load dataset
df = pd.read_csv(r"C:\Users\nohag\OneDrive\Desktop\UofT\12. Winter 2025\DSI\visualization\02_activities\assignments\neighbourhood-crime-rates - 4326.csv")
```

```python
# select one neighborhood to create the viz for
nbhd = 'Yonge-Bay Corridor'
row = df[df["AREA_NAME"]== nbhd]

# get columns for assault counts not assualt rates
assault_cols = [col for col in df.columns if col.startswith('ASSAULT_') and 'RATE' not in col]
assault_data = row[assault_cols].iloc[0]

# extract years and values
years = [int(col.split("_")[1]) for col in assault_cols]
values = assault_data.values
```

```python
# make the visualization

plt.figure(figsize=(10,6))
plt.plot(years, values, marker = 'o', linestyle='-', color = 'red')
plt.title('Assault Crimes over Time in Yonge-Bay Corridor')
plt.xlabel('Year')
plt.ylabel('Number of Assaults')
plt.grid(True)
plt.tight_layout()
plt.show()

# Add alt-text
print("Alt-text: Line chart showing assault crimes from 2014 to 2024 in Yonge-Bay Corridor. Assaults rose from 387 in 2014 to a peak of 660 in 2019, then declined to 377 in 2020.")
```
